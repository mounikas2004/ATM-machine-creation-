```python
import tkinter as tk
from tkinter import messagebox

# Demo account details
PIN = "1234"
balance = 10000.00
transactions = []

# ATM window
root = tk.Tk()
root.title("ATM Machine")
root.geometry("400x600")
root.resizable(False, False)
root.configure(bg="#102a43")

current_pin = PIN


def clear_screen():
    for widget in root.winfo_children():
        widget.destroy()


def make_label(text, size=16, color="white"):
    return tk.Label(
        root,
        text=text,
        font=("Arial", size, "bold"),
        bg="#102a43",
        fg=color
    )


def make_button(text, command, color="#168aad"):
    return tk.Button(
        root,
        text=text,
        command=command,
        font=("Arial", 12, "bold"),
        bg=color,
        fg="white",
        activebackground="#34a0a4",
        relief="flat",
        width=22,
        height=2
    )


def show_login():
    clear_screen()

    make_label("WELCOME TO PYTHON ATM", 19).pack(pady=35)
    make_label("Enter your 4-digit PIN").pack(pady=10)

    pin_entry = tk.Entry(
        root, show="*", font=("Arial", 20),
        justify="center", width=12
    )
    pin_entry.pack(pady=10)
    pin_entry.focus_set()

    def login():
        if pin_entry.get() == current_pin:
            show_menu()
        else:
            messagebox.showerror("ATM", "Incorrect PIN!")
            pin_entry.delete(0, tk.END)

    make_button("LOGIN", login).pack(pady=15)
    pin_entry.bind("<Return>", lambda event: login())
    make_label("Demo PIN: 1234", 10, "#bcccdc").pack(pady=10)


def show_menu():
    clear_screen()

    make_label("ATM MAIN MENU", 22).pack(pady=30)

    make_button("Check Balance", check_balance).pack(pady=8)
    make_button("Withdraw Cash", withdraw_cash).pack(pady=8)
    make_button("Deposit Cash", deposit_cash).pack(pady=8)
    make_button("Change PIN", change_pin).pack(pady=8)
    make_button("Mini Statement", mini_statement).pack(pady=8)
    make_button("Exit", show_login, "#b83232").pack(pady=8)


def check_balance():
    messagebox.showinfo(
        "Account Balance",
        f"Your balance is ₹{balance:,.2f}"
    )


def withdraw_cash():
    global balance

    window = tk.Toplevel(root)
    window.title("Withdraw Cash")
    window.geometry("300x200")
    window.configure(bg="#102a43")

    tk.Label(
        window, text="Enter amount (₹)",
        font=("Arial", 14), bg="#102a43", fg="white"
    ).pack(pady=15)

    amount_entry = tk.Entry(window, font=("Arial", 16))
    amount_entry.pack(pady=5)
    amount_entry.focus_set()

    def withdraw():
        global balance

        try:
            amount = float(amount_entry.get())

            if amount <= 0:
                messagebox.showerror("Error", "Enter a positive amount.")
            elif amount > balance:
                messagebox.showerror("Error", "Insufficient balance.")
            else:
                balance -= amount
                transactions.append(f"Withdraw: -₹{amount:.2f}")
                messagebox.showinfo(
                    "Success",
                    f"Please collect your cash.\n"
                    f"Remaining balance: ₹{balance:,.2f}"
                )
                window.destroy()
        except ValueError:
            messagebox.showerror("Error", "Enter a valid amount.")

    tk.Button(
        window, text="WITHDRAW", command=withdraw,
        font=("Arial", 12), bg="#168aad", fg="white"
    ).pack(pady=12)


def deposit_cash():
    global balance

    window = tk.Toplevel(root)
    window.title("Deposit Cash")
    window.geometry("300x200")
    window.configure(bg="#102a43")

    tk.Label(
        window, text="Enter deposit amount (₹)",
        font=("Arial", 12), bg="#102a43", fg="white"
    ).pack(pady=15)

    amount_entry = tk.Entry(window, font=("Arial", 16))
    amount_entry.pack(pady=5)
    amount_entry.focus_set()

    def deposit():
        global balance

        try:
            amount = float(amount_entry.get())

            if amount <= 0:
                messagebox.showerror("Error", "Enter a positive amount.")
            else:
                balance += amount
                transactions.append(f"Deposit: +₹{amount:.2f}")
                messagebox.showinfo(
                    "Success",
                    f"Deposit successful!\n"
                    f"New balance: ₹{balance:,.2f}"
                )
                window.destroy()
        except ValueError:
            messagebox.showerror("Error", "Enter a valid amount.")

    tk.Button(
        window, text="DEPOSIT", command=deposit,
        font=("Arial", 12), bg="#168aad", fg="white"
    ).pack(pady=12)


def change_pin():
    global current_pin

    window = tk.Toplevel(root)
    window.title("Change PIN")
    window.geometry("300x250")
    window.configure(bg="#102a43")

    tk.Label(
        window, text="Enter current PIN",
        bg="#102a43", fg="white"
    ).pack(pady=8)

    old_entry = tk.Entry(window, show="*", font=("Arial", 14))
    old_entry.pack()

    tk.Label(
        window, text="Enter new 4-digit PIN",
        bg="#102a43", fg="white"
    ).pack(pady=8)

    new_entry = tk.Entry(window, show="*", font=("Arial", 14))
    new_entry.pack()

    def update_pin():
        global current_pin

        old_pin = old_entry.get()
        new_pin = new_entry.get()

        if old_pin != current_pin:
            messagebox.showerror("Error", "Incorrect current PIN.")
        elif len(new_pin) != 4 or not new_pin.isdigit():
            messagebox.showerror(
                "Error", "PIN must contain exactly 4 digits."
            )
        else:
            current_pin = new_pin
            messagebox.showinfo("Success", "PIN changed successfully.")
            window.destroy()

    tk.Button(
        window, text="CHANGE PIN", command=update_pin,
        bg="#168aad", fg="white"
    ).pack(pady=15)


def mini_statement():
    if transactions:
        statement = "\n".join(transactions[-10:])
    else:
        statement = "No transactions yet."

    messagebox.showinfo("Mini Statement", statement)


# Start ATM application
show_login()
root.mainloop()

import tkinter as tk
from tkinter import messagebox

# Demo account details
PIN = "1234"
balance = 10000.00
transactions = []

# ATM window
root = tk.Tk()
root.title("ATM Machine")
root.geometry("400x600")
root.resizable(False, False)
root.configure(bg="#102a43")

current_pin = PIN


def clear_screen():
    for widget in root.winfo_children():
        widget.destroy()


def make_label(text, size=16, color="white"):
    return tk.Label(
        root,
        text=text,
        font=("Arial", size, "bold"),
        bg="#102a43",
        fg=color
    )


def make_button(text, command, color="#168aad"):
    return tk.Button(
        root,
        text=text,
        command=command,
        font=("Arial", 12, "bold"),
        bg=color,
        fg="white",
        activebackground="#34a0a4",
        relief="flat",
        width=22,
        height=2
    )


def show_login():
    clear_screen()

    make_label("WELCOME TO PYTHON ATM", 19).pack(pady=35)
    make_label("Enter your 4-digit PIN").pack(pady=10)

    pin_entry = tk.Entry(
        root, show="*", font=("Arial", 20),
        justify="center", width=12
    )
    pin_entry.pack(pady=10)
    pin_entry.focus_set()

    def login():
        if pin_entry.get() == current_pin:
            show_menu()
        else:
            messagebox.showerror("ATM", "Incorrect PIN!")
            pin_entry.delete(0, tk.END)

    make_button("LOGIN", login).pack(pady=15)
    pin_entry.bind("<Return>", lambda event: login())
    make_label("Demo PIN: 1234", 10, "#bcccdc").pack(pady=10)


def show_menu():
    clear_screen()

    make_label("ATM MAIN MENU", 22).pack(pady=30)

    make_button("Check Balance", check_balance).pack(pady=8)
    make_button("Withdraw Cash", withdraw_cash).pack(pady=8)
    make_button("Deposit Cash", deposit_cash).pack(pady=8)
    make_button("Change PIN", change_pin).pack(pady=8)
    make_button("Mini Statement", mini_statement).pack(pady=8)
    make_button("Exit", show_login, "#b83232").pack(pady=8)


def check_balance():
    messagebox.showinfo(
        "Account Balance",
        f"Your balance is ₹{balance:,.2f}"
    )


def withdraw_cash():
    global balance

    window = tk.Toplevel(root)
    window.title("Withdraw Cash")
    window.geometry("300x200")
    window.configure(bg="#102a43")

    tk.Label(
        window, text="Enter amount (₹)",
        font=("Arial", 14), bg="#102a43", fg="white"
    ).pack(pady=15)

    amount_entry = tk.Entry(window, font=("Arial", 16))
    amount_entry.pack(pady=5)
    amount_entry.focus_set()

    def withdraw():
        global balance

        try:
            amount = float(amount_entry.get())

            if amount <= 0:
                messagebox.showerror("Error", "Enter a positive amount.")
            elif amount > balance:
                messagebox.showerror("Error", "Insufficient balance.")
            else:
                balance -= amount
                transactions.append(f"Withdraw: -₹{amount:.2f}")
                messagebox.showinfo(
                    "Success",
                    f"Please collect your cash.\n"
                    f"Remaining balance: ₹{balance:,.2f}"
                )
                window.destroy()
        except ValueError:
            messagebox.showerror("Error", "Enter a valid amount.")

    tk.Button(
        window, text="WITHDRAW", command=withdraw,
        font=("Arial", 12), bg="#168aad", fg="white"
    ).pack(pady=12)


def deposit_cash():
    global balance

    window = tk.Toplevel(root)
    window.title("Deposit Cash")
    window.geometry("300x200")
    window.configure(bg="#102a43")

    tk.Label(
        window, text="Enter deposit amount (₹)",
        font=("Arial", 12), bg="#102a43", fg="white"
    ).pack(pady=15)

    amount_entry = tk.Entry(window, font=("Arial", 16))
    amount_entry.pack(pady=5)
    amount_entry.focus_set()

    def deposit():
        global balance

        try:
            amount = float(amount_entry.get())

            if amount <= 0:
                messagebox.showerror("Error", "Enter a positive amount.")
            else:
                balance += amount
                transactions.append(f"Deposit: +₹{amount:.2f}")
                messagebox.showinfo(
                    "Success",
                    f"Deposit successful!\n"
                    f"New balance: ₹{balance:,.2f}"
                )
                window.destroy()
        except ValueError:
            messagebox.showerror("Error", "Enter a valid amount.")

    tk.Button(
        window, text="DEPOSIT", command=deposit,
        font=("Arial", 12), bg="#168aad", fg="white"
    ).pack(pady=12)


def change_pin():
    global current_pin

    window = tk.Toplevel(root)
    window.title("Change PIN")
    window.geometry("300x250")
    window.configure(bg="#102a43")

    tk.Label(
        window, text="Enter current PIN",
        bg="#102a43", fg="white"
    ).pack(pady=8)

    old_entry = tk.Entry(window, show="*", font=("Arial", 14))
    old_entry.pack()

    tk.Label(
        window, text="Enter new 4-digit PIN",
        bg="#102a43", fg="white"
    ).pack(pady=8)

    new_entry = tk.Entry(window, show="*", font=("Arial", 14))
    new_entry.pack()

    def update_pin():
        global current_pin

        old_pin = old_entry.get()
        new_pin = new_entry.get()

        if old_pin != current_pin:
            messagebox.showerror("Error", "Incorrect current PIN.")
        elif len(new_pin) != 4 or not new_pin.isdigit():
            messagebox.showerror(
                "Error", "PIN must contain exactly 4 digits."
            )
        else:
            current_pin = new_pin
            messagebox.showinfo("Success", "PIN changed successfully.")
            window.destroy()

    tk.Button(
        window, text="CHANGE PIN", command=update_pin,
        bg="#168aad", fg="white"
    ).pack(pady=15)


def mini_statement():
    if transactions:
        statement = "\n".join(transactions[-10:])
    else:
        statement = "No transactions yet."

    messagebox.showinfo("Mini Statement", statement)


# Start ATM application
show_login()
root.mainloop()

import tkinter as tk
from tkinter import messagebox

# Demo account details
PIN = "1234"
balance = 10000.00
transactions = []

# ATM window
root = tk.Tk()
root.title("ATM Machine")
root.geometry("400x600")
root.resizable(False, False)
root.configure(bg="#102a43")

current_pin = PIN


def clear_screen():
    for widget in root.winfo_children():
        widget.destroy()


def make_label(text, size=16, color="white"):
    return tk.Label(
        root,
        text=text,
        font=("Arial", size, "bold"),
        bg="#102a43",
        fg=color
    )


def make_button(text, command, color="#168aad"):
    return tk.Button(
        root,
        text=text,
        command=command,
        font=("Arial", 12, "bold"),
        bg=color,
        fg="white",
        activebackground="#34a0a4",
        relief="flat",
        width=22,
        height=2
    )


def show_login():
    clear_screen()

    make_label("WELCOME TO PYTHON ATM", 19).pack(pady=35)
    make_label("Enter your 4-digit PIN").pack(pady=10)

    pin_entry = tk.Entry(
        root, show="*", font=("Arial", 20),
        justify="center", width=12
    )
    pin_entry.pack(pady=10)
    pin_entry.focus_set()

    def login():
        if pin_entry.get() == current_pin:
            show_menu()
        else:
            messagebox.showerror("ATM", "Incorrect PIN!")
            pin_entry.delete(0, tk.END)

    make_button("LOGIN", login).pack(pady=15)
    pin_entry.bind("<Return>", lambda event: login())
    make_label("Demo PIN: 1234", 10, "#bcccdc").pack(pady=10)


def show_menu():
    clear_screen()

    make_label("ATM MAIN MENU", 22).pack(pady=30)

    make_button("Check Balance", check_balance).pack(pady=8)
    make_button("Withdraw Cash", withdraw_cash).pack(pady=8)
    make_button("Deposit Cash", deposit_cash).pack(pady=8)
    make_button("Change PIN", change_pin).pack(pady=8)
    make_button("Mini Statement", mini_statement).pack(pady=8)
    make_button("Exit", show_login, "#b83232").pack(pady=8)


def check_balance():
    messagebox.showinfo(
        "Account Balance",
        f"Your balance is ₹{balance:,.2f}"
    )


def withdraw_cash():
    global balance

    window = tk.Toplevel(root)
    window.title("Withdraw Cash")
    window.geometry("300x200")
    window.configure(bg="#102a43")

    tk.Label(
        window, text="Enter amount (₹)",
        font=("Arial", 14), bg="#102a43", fg="white"
    ).pack(pady=15)

    amount_entry = tk.Entry(window, font=("Arial", 16))
    amount_entry.pack(pady=5)
    amount_entry.focus_set()

    def withdraw():
        global balance

        try:
            amount = float(amount_entry.get())

            if amount <= 0:
                messagebox.showerror("Error", "Enter a positive amount.")
            elif amount > balance:
                messagebox.showerror("Error", "Insufficient balance.")
            else:
                balance -= amount
                transactions.append(f"Withdraw: -₹{amount:.2f}")
                messagebox.showinfo(
                    "Success",
                    f"Please collect your cash.\n"
                    f"Remaining balance: ₹{balance:,.2f}"
                )
                window.destroy()
        except ValueError:
            messagebox.showerror("Error", "Enter a valid amount.")

    tk.Button(
        window, text="WITHDRAW", command=withdraw,
        font=("Arial", 12), bg="#168aad", fg="white"
    ).pack(pady=12)


def deposit_cash():
    global balance

    window = tk.Toplevel(root)
    window.title("Deposit Cash")
    window.geometry("300x200")
    window.configure(bg="#102a43")

    tk.Label(
        window, text="Enter deposit amount (₹)",
        font=("Arial", 12), bg="#102a43", fg="white"
    ).pack(pady=15)

    amount_entry = tk.Entry(window, font=("Arial", 16))
    amount_entry.pack(pady=5)
    amount_entry.focus_set()

    def deposit():
        global balance

        try:
            amount = float(amount_entry.get())

            if amount <= 0:
                messagebox.showerror("Error", "Enter a positive amount.")
            else:
                balance += amount
                transactions.append(f"Deposit: +₹{amount:.2f}")
                messagebox.showinfo(
                    "Success",
                    f"Deposit successful!\n"
                    f"New balance: ₹{balance:,.2f}"
                )
                window.destroy()
        except ValueError:
            messagebox.showerror("Error", "Enter a valid amount.")

    tk.Button(
        window, text="DEPOSIT", command=deposit,
        font=("Arial", 12), bg="#168aad", fg="white"
    ).pack(pady=12)


def change_pin():
    global current_pin

    window = tk.Toplevel(root)
    window.title("Change PIN")
    window.geometry("300x250")
    window.configure(bg="#102a43")

    tk.Label(
        window, text="Enter current PIN",
        bg="#102a43", fg="white"
    ).pack(pady=8)

    old_entry = tk.Entry(window, show="*", font=("Arial", 14))
    old_entry.pack()

    tk.Label(
        window, text="Enter new 4-digit PIN",
        bg="#102a43", fg="white"
    ).pack(pady=8)

    new_entry = tk.Entry(window, show="*", font=("Arial", 14))
    new_entry.pack()

    def update_pin():
        global current_pin

        old_pin = old_entry.get()
        new_pin = new_entry.get()

        if old_pin != current_pin:
            messagebox.showerror("Error", "Incorrect current PIN.")
        elif len(new_pin) != 4 or not new_pin.isdigit():
            messagebox.showerror(
                "Error", "PIN must contain exactly 4 digits."
            )
        else:
            current_pin = new_pin
            messagebox.showinfo("Success", "PIN changed successfully.")
            window.destroy()

    tk.Button(
        window, text="CHANGE PIN", command=update_pin,
        bg="#168aad", fg="white"
    ).pack(pady=15)


def mini_statement():
    if transactions:
        statement = "\n".join(transactions[-10:])
    else:
        statement = "No transactions yet."

    messagebox.showinfo("Mini Statement", statement)


# Start ATM application
show_login()
root.mainloop()
```
